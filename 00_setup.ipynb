```python
CATALOG = "opspulse"
SCHEMAS = ["bronze", "silver", "gold"]
VOLUME_SCHEMA = "bronze"
VOLUME_NAME = "landing"
LANDING_PATH = f"/Volumes/{CATALOG}/{VOLUME_SCHEMA}/{VOLUME_NAME}"
print(LANDING_PATH)
```

```python
import logging
logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
log = logging.getLogger("opspulse.setup")

try:
    spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
    for s in SCHEMAS:
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{s}")
        log.info("schema ready: %s.%s", CATALOG, s)
    spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{VOLUME_SCHEMA}.{VOLUME_NAME}")
    log.info("volume ready: %s", LANDING_PATH)
except Exception as e:
    log.error("Setup failed: %s", e)
    raise
```

```python
display(spark.sql(f"SHOW SCHEMAS IN {CATALOG}"))
display(spark.sql(f"SHOW VOLUMES IN {CATALOG}.{VOLUME_SCHEMA}"))
```
